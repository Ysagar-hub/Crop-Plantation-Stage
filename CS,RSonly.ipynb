```python
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]
print('Uploaded:', filename)
```

```python
import pandas as pd

df = pd.read_excel(filename, sheet_name='Compiled', header=1)
df.head(10)
```

```python
df_clean = df[df['District'].notna() & (df['District'] != 'Grand Total')].copy()

df_clean['has_CS'] = df_clean['CS'].notna()
df_clean['has_RS'] = df_clean['RS'].notna()
df_clean['has_SS'] = df_clean['SS'].notna()

df_clean[['District', 'has_CS', 'has_RS', 'has_SS']]
```

```python
all_three = df_clean[df_clean['has_CS'] & df_clean['has_RS'] & df_clean['has_SS']]['District'].tolist()
cs_rs_only = df_clean[df_clean['has_CS'] & df_clean['has_RS'] & ~df_clean['has_SS']]['District'].tolist()

print('Districts with CS + RS + SS (all three):')
for d in all_three:
    print(' -', d)

print('\nDistricts with only CS + RS (no SS):')
for d in cs_rs_only:
    print(' -', d)
```
